# Lab 2: Data Preprocessing using Pandas

### Objectives:
 1. Import data and perform exploratory checks.
 2. Handle missing values using mean and mode imputation.
 3. Remove duplicate records from the dataset.
 4. Encode categorical variables into numerical formats.
 5. Apply Normalization and Standardization.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler,MinMaxScaler
from sklearn.preprocessing import LabelEncoder

df=pd.read_csv("/content/synthetic_dataset.csv")
```

```python
df.info()
df.shape
```

```python
df = df.drop_duplicates()
print(f"Dataset shape after removing duplicates: {df.shape}")
```

```python
df.isna().sum()
```

```python
df['Category']=df['Category'].fillna(df['Category'].mode()[0])
df['Price']=df['Price'].fillna(df['Price'].mean())
df['Rating']=df['Rating'].fillna(df['Rating'].mean())
df['Stock']=df['Stock'].fillna(df['Stock'].mode()[0])
df['Discount']=df['Discount'].fillna(df['Discount'].mean())
```

```python
df.isna().sum()
```

```python
df.describe()
```

```python
le = LabelEncoder()
df['Category'] = le.fit_transform(df['Category'])
df['Stock'] = le.fit_transform(df['Stock'])
df.head()
```

```python
df.describe()
```

```python
scaler = StandardScaler()
numerical_cols = ['Price', 'Rating', 'Discount']
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

```python
df.describe()
```

```python
scaler = MinMaxScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
df.describe()
df.head(10)
```
